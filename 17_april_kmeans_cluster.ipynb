```python
import pandas as pd
from sklearn.cluster import KMeans
from matplotlib import pyplot as plt
import numpy as np
```

```python
data=pd.DataFrame({'x':[2,4,10,12,3,20,30,11,25,7]})
kmeans=KMeans(n_clusters=2,max_iter=100)
kmeans.fit(data)
print('\n Sum of square error/ WCSS is: ',kmeans.inertia_)

data['labels']=kmeans.labels_
data
```

```python
k_values=[]
wcss_values=[]
for k in range(1,10):
    kmeans=KMeans(n_clusters=k,max_iter=100)
    kmeans.fit(data)
    k_values.append(k)
    wcss_values.append(kmeans.inertia_)

plt.plot(k_values,wcss_values)
plt.show()
```

```python
x=df.drop('Species',axis=1)
y = df['Species']
```

```python
from sklearn.preprocessing import StandardScaler

scaler=StandardScaler()
x_scaled=scaler.fit_transform(x)
```

```python
x
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x_scaled, y, test_size=0.3, random_state=1)
```

```python
from sklearn.cluster import KMeans
model=KMeans(n_clusters=3)
```

```python
model.fit(X_train)
```

```python
y_pred=model.predict(X_test)
```

```python
from collections import defaultdict

clusters_to_species={}
for i in range(3):
    ci=(model.labels_==i)
    mcs=y_train[ci].value_counts().index[0]
    cluster_to_species[i]=mcs
```

```python
y_pred_species=[cluster_to_species[cluster] for cluster in y_pred]
```

```python
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score

print(confusion_matrix(y_test, y_pred_species))
```

```python
print(classification_report(y_test, y_pred_species))
print(accuracy_score(y_test, y_pred_species))
```
